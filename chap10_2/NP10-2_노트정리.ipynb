```python
import socket
#문자열 이름
HOSTS = [
'www.dongyang.ac.kr', 'pymotw.com','www.seowon.ac.kr',
'www.python.org',
'testsite',
]
for host in HOSTS:
    try:
        print('{} : {}'.format(host, socket.gethostbyname(host)))
    except socket.error as msg: 
        print('{} : {}'.format(host, msg))
```
